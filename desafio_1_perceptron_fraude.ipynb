```python
import numpy as np
```

```python
# Dados de treinamento
X_treino_fraude = np.array([
    [1.5, 1.0],
    [2.0, 2.0],
    [3.5, 1.5],
    [3.0, 3.0],
    [6.5, 5.0],
    [7.0, 7.0],
    [8.5, 6.0],
    [9.0, 8.0]
])

# 0 = Legítima | 1 = Suspeita
y_treino_fraude = np.array([
    0, 0, 0, 0,
    1, 1, 1, 1
])

print("=== DADOS DE TREINAMENTO ===")
print(X_treino_fraude)

print("\n=== RÓTULOS ===")
print(y_treino_fraude)
```

```python
# Função de treinamento do Perceptron
def treinar_perceptron(X, y, taxa_aprendizado, epocas):

    pesos = np.ones(X.shape[1])
    vies = 1.0

    for epoca in range(epocas):

        erros = 0

        for i in range(len(X)):

            soma = np.dot(X[i], pesos) + vies
            previsao = 1 if soma >= 0 else 0
            erro = y[i] - previsao

            if erro != 0:

                # Regra de atualização de Rosenblatt
                pesos = pesos + taxa_aprendizado * erro * X[i]
                vies = vies + taxa_aprendizado * erro

                erros += 1

        if erros == 0:
            break

    return pesos, vies
```

```python
# Configuração do treinamento
taxa_aprendizado = 0.1
epocas = 100

pesos_finais, vies_final = treinar_perceptron(
    X_treino_fraude,
    y_treino_fraude,
    taxa_aprendizado,
    epocas
)

print("=== PARÂMETROS FINAIS ===")
print(f"Pesos: {pesos_finais}")
print(f"Viés: {vies_final}")
```

```python
# Função de predição
def prever(amostra, pesos, vies):

    soma = np.dot(amostra, pesos) + vies

    # Função degrau
    return 1 if soma >= 0 else 0
```

```python
# Novas transações
transacao_A = np.array([5.0, 4.0])
transacao_B = np.array([7.0, 6.5])

previsao_A = prever(
    transacao_A,
    pesos_finais,
    vies_final
)

previsao_B = prever(
    transacao_B,
    pesos_finais,
    vies_final
)
```

```python
# Função para exibir o diagnóstico
def diagnostico(classe):

    if classe == 0:
        return "Transação Legítima"

    return "Transação Suspeita (Risco de Fraude)"


print("=== DIAGNÓSTICO FINAL ===")
print(f"Transação A: {diagnostico(previsao_A)}")
print(f"Transação B: {diagnostico(previsao_B)}")
```
