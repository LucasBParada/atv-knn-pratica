```python
import numpy as np
```

```python
# Dados de treinamento
# [Dias de Inatividade, Chamados Críticos]
X_treino_churn = np.array([
    [2.0, 1.0],
    [3.0, 0.0],
    [5.0, 1.0],
    [6.0, 2.0],
    [15.0, 4.0],
    [18.0, 5.0],
    [20.0, 4.0],
    [22.0, 6.0]
])

# 0 = Baixo Risco | 1 = Alto Risco
y_treino_churn = np.array([
    0, 0, 0, 0,
    1, 1, 1, 1
])

print("=== DADOS DE TREINAMENTO ===")
print(X_treino_churn)

print("\n=== RÓTULOS ===")
print(y_treino_churn)
```

```python
# Função de classificação KNN
def classificar_knn(X, y, ponto, k, metrica="euclidiana"):

    # Calcula as distâncias de forma vetorizada
    if metrica == "euclidiana":

        distancias = np.sqrt(
            np.sum((X - ponto) ** 2, axis=1)
        )

    elif metrica == "manhattan":

        distancias = np.sum(
            np.abs(X - ponto),
            axis=1
        )

    else:
        raise ValueError(
            "Métrica deve ser 'euclidiana' ou 'manhattan'"
        )

    # Seleciona os k vizinhos mais próximos
    indices_vizinhos = np.argsort(distancias)[:k]

    # Obtém as classes dos vizinhos
    classes_vizinhos = y[indices_vizinhos]

    # Define a classe pela maioria
    classes, contagens = np.unique(
        classes_vizinhos,
        return_counts=True
    )

    classe_predita = classes[np.argmax(contagens)]

    return (
        classe_predita,
        indices_vizinhos,
        distancias[indices_vizinhos]
    )
```

```python
# Clientes para classificação
cliente_1 = np.array([13.0, 10.0])
cliente_2 = np.array([7.0, 1.5])

# Configuração do KNN
k = 3
metrica = "euclidiana"
```

```python
# Classifica os clientes
classe_1, vizinhos_1, distancias_1 = classificar_knn(
    X_treino_churn,
    y_treino_churn,
    cliente_1,
    k,
    metrica
)

classe_2, vizinhos_2, distancias_2 = classificar_knn(
    X_treino_churn,
    y_treino_churn,
    cliente_2,
    k,
    metrica
)
```

```python
# Converte a classe em diagnóstico
def diagnostico(classe):

    if classe == 0:
        return "Baixo Risco"

    return "Alto Risco"
```

```python
print("=== DIAGNÓSTICO DOS CLIENTES ===")

print("\nCliente 1")
print(f"Classe: {diagnostico(classe_1)}")
print(f"Índices dos vizinhos: {vizinhos_1}")
print(f"Distâncias: {distancias_1}")

print("\nCliente 2")
print(f"Classe: {diagnostico(classe_2)}")
print(f"Índices dos vizinhos: {vizinhos_2}")
print(f"Distâncias: {distancias_2}")
```
