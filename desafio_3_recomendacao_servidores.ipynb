```python
import numpy as np
```

```python
# Catálogo
# [vCPUs, RAM (GB), SSD (GB)]
catalogo_servidores = np.array([
    [2.0, 4.0, 50.0],
    [4.0, 8.0, 100.0],
    [8.0, 16.0, 250.0],
    [16.0, 32.0, 500.0],
    [32.0, 64.0, 1000.0],
    [64.0, 128.0, 2000.0]
])

# Nomes comerciais
nomes_servidores = [
    "Micro Instância Web",
    "Standard App Server",
    "Medium Backend & Cache",
    "Database Enterprise",
    "High Performance Computing",
    "Big Data & AI Training"
]

print("=== CATÁLOGO DE SERVIDORES ===")
print(catalogo_servidores)
```

```python
# Requisitos da aplicação
perfil_demandado = np.array([12.0, 28.0, 850.0])

# Quantidade de recomendações
k = 2
```

```python
# Função de recomendação
def recomendar_servidores(catalogo, perfil, k):

    # Calcula as distâncias de forma vetorizada
    distancias = np.sqrt(
        np.sum((catalogo - perfil) ** 2, axis=1)
    )

    # Ordena pela menor distância
    indices_ordenados = np.argsort(distancias)

    # Seleciona as melhores recomendações
    indices_recomendados = indices_ordenados[:k]

    return indices_recomendados, distancias[indices_recomendados]
```

```python
# Executa a recomendação
indices_recomendados, distancias_recomendadas = recomendar_servidores(
    catalogo_servidores,
    perfil_demandado,
    k
)
```

```python
# Exibe o ranking
print("=== RANKING DE SERVIDORES RECOMENDADOS ===")

for posicao, (indice, distancia) in enumerate(
    zip(indices_recomendados, distancias_recomendadas),
    start=1
):

    servidor = catalogo_servidores[indice]
    nome = nomes_servidores[indice]

    print(f"\n{posicao}º lugar - {nome}")
    print(f"vCPUs: {servidor[0]:.0f}")
    print(f"RAM: {servidor[1]:.0f} GB")
    print(f"SSD: {servidor[2]:.0f} GB")
    print(f"Distância: {distancia:.2f}")
```
